# Pareto: index size vs nDCG@10 on JQaRA

Reads `results/grid.csv` (produced by `make reproduce`) and redraws the charts. Run with `uv run --extra notebook jupyter lab`.

```python
import pandas as pd
from mrl_quant_ja.config import load_config
from mrl_quant_ja.report import _wide, find_knee, plot_pareto, config_name

cfg = load_config()
w = _wide(pd.read_csv(cfg['results_dir'] / 'grid.csv'))
knee = find_knee(w[w.model == 'eg2'])
print('knee:', config_name(knee))
```

```python
eg2 = w[w.model == 'eg2'].copy()
eg2['config'] = eg2.apply(config_name, axis=1)
eg2.sort_values('bytes_per_vector')[['config', 'bytes_per_vector', 'ndcg@10 full', 'recall@100 full', 'ndcg@10 rerank']]
```

```python
from IPython.display import Image
plot_pareto(w, knee, cfg['figures_dir'] / 'pareto.png', 'full')
Image(cfg['figures_dir'] / 'pareto.png')
```
